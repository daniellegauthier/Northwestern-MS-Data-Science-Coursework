# 🎨 Semantic Partial Derivative Explorer
This interactive notebook allows you to explore the semantic momentum of color sequences using partial derivatives over RGB dimensions. Each color channel is mapped to a conceptual axis:

Red → Time

Green → Space

Blue → Light

## Using numerical differentiation, the notebook estimates how small changes in each dimension affect the underlying scalar field—representing meaning, sentiment, or any other defined function f(R,G,B).

## ✨ Features:
Choose a predefined color sequence.

Approximate partial derivatives:

∂f/∂R (Time)

∂f/∂G (Space)

∂f/∂B (Light)

## Visualize semantic gradients over the course of a sequence.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, Dropdown

# Load data
sequences_df = pd.read_csv("la matrice sequences.csv")
colors_df = pd.read_csv("la matrice.csv")

# Build RGB map
rgb_map = {}
for _, row in colors_df.iterrows():
    rgb_map[row['color']] = (row['r'], row['g'], row['b'])

# Function to compute partial derivatives
def compute_partial_derivatives(rgb_values):
    rgb_array = np.array(rgb_values, dtype=float)
    partials = {
        '∂Time (Red)': np.gradient(rgb_array[:, 0]),
        '∂Space (Green)': np.gradient(rgb_array[:, 1]),
        '∂Light (Blue)': np.gradient(rgb_array[:, 2])
    }
    return partials

# Function to analyze a sequence
def analyze_sequence(sequence_name):
    row = sequences_df[sequences_df['name'] == sequence_name]
    if row.empty:
        print("Sequence not found.")
        return

    colors = row.iloc[0]['sequence'].split(',')
    rgb_values = [rgb_map[color.strip()] for color in colors if color.strip() in rgb_map]

    if not rgb_values:
        print("No valid colors found in sequence.")
        return

    rgb_array = np.array(rgb_values)
    partials = compute_partial_derivatives(rgb_values)

    fig, ax = plt.subplots(figsize=(10, 6))
    x = list(range(len(rgb_values)))
    ax.plot(x, partials['∂Time (Red)'], label='∂Time (Red)', color='red')
    ax.plot(x, partials['∂Space (Green)'], label='∂Space (Green)', color='green')
    ax.plot(x, partials['∂Light (Blue)'], label='∂Light (Blue)', color='blue')
    ax.set_title(f"Partial Derivatives over Sequence: {sequence_name}")
    ax.set_xlabel("Step")
    ax.set_ylabel("Change Rate")
    ax.legend()
    plt.grid(True)
    plt.show()

# Create dropdown for interaction
sequence_dropdown = Dropdown(
    options=sequences_df['name'].unique().tolist(),
    description='Sequence:',
    disabled=False,
)

interact(analyze_sequence, sequence_name=sequence_dropdown)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, Dropdown
import sympy as sp

# Load data
sequences_df = pd.read_csv("la matrice sequences.csv")
colors_df = pd.read_csv("la matrice.csv")

# Build RGB map
rgb_map = {
    row['color']: (row['r'], row['g'], row['b'])
    for _, row in colors_df.iterrows()
}

# Function to compute symbolic and numeric partial derivatives
def compute_symbolic_and_numeric_derivatives(rgb_values):
    t = sp.Symbol('t', real=True)
    N = len(rgb_values)
    t_vals = list(range(N))

    r_vals = [rgb[0] for rgb in rgb_values]
    g_vals = [rgb[1] for rgb in rgb_values]
    b_vals = [rgb[2] for rgb in rgb_values]

    # Fit symbolic polynomials (degree N-1 interpolation)
    r_poly = sp.interpolate(list(zip(t_vals, r_vals)), t)
    g_poly = sp.interpolate(list(zip(t_vals, g_vals)), t)
    b_poly = sp.interpolate(list(zip(t_vals, b_vals)), t)

    # Derivatives
    dr_dt = sp.diff(r_poly, t)
    dg_dt = sp.diff(g_poly, t)
    db_dt = sp.diff(b_poly, t)

    # Evaluate numerically at each t
    dr_vals = [float(dr_dt.subs(t, i)) for i in t_vals]
    dg_vals = [float(dg_dt.subs(t, i)) for i in t_vals]
    db_vals = [float(db_dt.subs(t, i)) for i in t_vals]

    return {
        '∂Time (Red)': dr_vals,
        '∂Space (Green)': dg_vals,
        '∂Light (Blue)': db_vals,
        't_vals': t_vals,
        'r_poly': r_poly,
        'g_poly': g_poly,
        'b_poly': b_poly
    }

# Function to analyze a sequence
def analyze_sequence(sequence_name):
    row = sequences_df[sequences_df['name'] == sequence_name]
    if row.empty:
        print("Sequence not found.")
        return

    colors = row.iloc[0]['sequence'].split(',')
    rgb_values = [rgb_map[color.strip()] for color in colors if color.strip() in rgb_map]

    if not rgb_values:
        print("No valid colors found in sequence.")
        return

    result = compute_symbolic_and_numeric_derivatives(rgb_values)

    # Plot
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(result['t_vals'], result['∂Time (Red)'], label='∂Time (Red)', color='red')
    ax.plot(result['t_vals'], result['∂Space (Green)'], label='∂Space (Green)', color='green')
    ax.plot(result['t_vals'], result['∂Light (Blue)'], label='∂Light (Blue)', color='blue')
    ax.set_title(f"Symbolic Partial Derivatives over Sequence: {sequence_name}")
    ax.set_xlabel("Step")
    ax.set_ylabel("Rate of Change")
    ax.legend()
    plt.grid(True)
    plt.show()

    # Print symbolic expressions
    print("Symbolic RGB polynomials:")
    print(f"Red(t): {result['r_poly']}")
    print(f"Green(t): {result['g_poly']}")
    print(f"Blue(t): {result['b_poly']}\n")

# Interactive dropdown
sequence_dropdown = Dropdown(
    options=sequences_df['name'].unique().tolist(),
    description='Sequence:',
    layout={'width': '400px'},
)

interact(analyze_sequence, sequence_name=sequence_dropdown)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, Dropdown

# Load data
sequences_df = pd.read_csv("la matrice sequences.csv")
colors_df = pd.read_csv("la matrice.csv")

# Build RGB map
rgb_map = {
    row['color']: (row['r'], row['g'], row['b'])
    for _, row in colors_df.iterrows()
}

# Compute partial derivatives using finite differences
# Partial derivative with respect to red (time)
def partial_derivative_r(f, r, g, b, h=1e-3):
    return (f(r + h, g, b) - f(r, g, b)) / h

# Partial derivative with respect to green (space)
def partial_derivative_g(f, r, g, b, h=1e-3):
    return (f(r, g + h, b) - f(r, g, b)) / h

# Partial derivative with respect to blue (light)
def partial_derivative_b(f, r, g, b, h=1e-3):
    return (f(r, g, b + h) - f(r, g, b)) / h

# Function to analyze a sequence
def analyze_sequence(sequence_name):
    row = sequences_df[sequences_df['name'] == sequence_name]
    if row.empty:
        print("Sequence not found.")
        return

    colors = row.iloc[0]['sequence'].split(',')
    rgb_values = [rgb_map[color.strip()] for color in colors if color.strip() in rgb_map]

    if not rgb_values:
        print("No valid colors found in sequence.")
        return

    rgb_array = np.array(rgb_values, dtype=float)

    # Define a mock function to simulate semantic value from RGB (e.g., magnitude)
    def semantic_function(r, g, b):
        return r**2 + g**2 + b**2  # You can replace this with real sentiment or meaning scores

    # Compute partials
    partials_r = [partial_derivative_r(semantic_function, *rgb) for rgb in rgb_array]
    partials_g = [partial_derivative_g(semantic_function, *rgb) for rgb in rgb_array]
    partials_b = [partial_derivative_b(semantic_function, *rgb) for rgb in rgb_array]

    # Plot
    fig, ax = plt.subplots(figsize=(10, 6))
    x = list(range(len(rgb_values)))
    ax.plot(x, partials_r, label='∂Time (Red)', color='red')
    ax.plot(x, partials_g, label='∂Space (Green)', color='green')
    ax.plot(x, partials_b, label='∂Light (Blue)', color='blue')
    ax.set_title(f"Partial Derivatives over Sequence: {sequence_name}")
    ax.set_xlabel("Step")
    ax.set_ylabel("Change Rate")
    ax.legend()
    plt.grid(True)
    plt.show()

# Create dropdown for interaction
sequence_dropdown = Dropdown(
    options=sequences_df['name'].unique().tolist(),
    description='Sequence:',
    disabled=False,
)

interact(analyze_sequence, sequence_name=sequence_dropdown)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objs as go
from ipywidgets import interact, Dropdown, Text
from IPython.display import display

# Load the semantic color data
df = pd.read_csv("semantic_rgb_mapping_with_sentiment.csv")

# Normalize color names for comparison
df['Normalized Color'] = df['Closest Color'].str.strip().str.lower()

# Assign clusters manually based on pre-analyzed clusters
cluster_map = {
    'gold': 0, 'yellow': 0,
    'grey': 1, 'nude': 1, 'white': 1,
    'pink': 2, 'orange': 2, 'red': 2, 'brown': 2,
    'blue': 3, 'purple': 3,
    'green': 4, 'black': 4
}
df['Cluster'] = df['Normalized Color'].map(cluster_map)

# Get unique normalized colors and their average RGB/sentiment
color_summary = df.groupby(['Normalized Color', 'Cluster']).agg({
    'R': 'mean', 'G': 'mean', 'B': 'mean',
    'Sentiment Score': 'mean',
    'Original Words': lambda x: x.iloc[0]
}).reset_index()

# Assign cluster colors for visualization
cluster_colors = ['#FFD700', '#D3D3D3', '#FF6347', '#9370DB', '#228B22']
color_summary['Cluster Color'] = color_summary['Cluster'].apply(
    lambda c: cluster_colors[int(c)] if pd.notnull(c) else '#CCCCCC'
)

# Prepare the 3D cube data with cluster coloring
trace = go.Scatter3d(
    x=color_summary['R'],
    y=color_summary['G'],
    z=color_summary['B'],
    mode='markers+text',
    marker=dict(
        size=6,
        color=color_summary['Cluster Color'],
        opacity=0.9,
    ),
    text=color_summary['Normalized Color'] + '<br>' +
         'Semantic: ' + color_summary['Original Words'].astype(str) + '<br>' +
         'Sentiment: ' + color_summary['Sentiment Score'].round(2).astype(str) + '<br>' +
         'Cluster: ' + color_summary['Cluster'].astype(str),
    hoverinfo='text'
)

layout = go.Layout(
    title='🧠 Semantic RGB Cube with Clusters and Sentiment',
    scene=dict(
        xaxis=dict(title='Red (Time)'),
        yaxis=dict(title='Green (Space)'),
        zaxis=dict(title='Blue (Light)')
    ),
    height=700
)

fig = go.Figure(data=[trace], layout=layout)
fig.show()

# Add interactive search widget
def search_color(color_name):
    color_name = color_name.strip().lower()
    match = color_summary[color_summary['Normalized Color'] == color_name]
    if not match.empty:
        print(f"🎨 Color: {color_name.title()}")
        print(f"RGB: ({int(match['R'].values[0])}, {int(match['G'].values[0])}, {int(match['B'].values[0])})")
        print(f"Sentiment: {match['Sentiment Score'].values[0]:.3f}")
        print(f"Top Semantic Words: {match['Original Words'].values[0]}")
        print(f"Cluster: {int(match['Cluster'].values[0])}")
    else:
        print("No exact match found. Try a simpler color name.")

interact(search_color, color_name=Text(value='yellow', description='Search:'))
```
