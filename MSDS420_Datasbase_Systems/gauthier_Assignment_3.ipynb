### Requirement 6:
- List the top 10 distribution centers that have **highest** number of delayed deliveries on sales (not returns)? Show the distribution center ID, the city and the number of delayed deliveries. **( Use Python - SQL - Postgres )**




```python
import pandas as pd

sql = """
    SELECT 
    t.deliveryzipcode, 
    l.city,
    l.distribution_center_id,
    COUNT(t.orderid) AS delayed_orders,
    SUM(t.actualdeliverydate::date - t.expecteddeliverydate::date) AS total_delay,
    ROUND(AVG(t.actualdeliverydate::date - t.expecteddeliverydate::date), 2) AS delay_days
FROM transactions_log t
JOIN logistics_supply_chain_network l
    ON t.deliveryzipcode = l.zipcode
WHERE orderreturned = 'No' 
GROUP BY t.deliveryzipcode, l.city, l.distribution_center_id
ORDER BY delayed_orders DESC 
LIMIT 10;
"""

# Load directly into a DataFrame using your active psycopg2 connection ('conn')
df6 = pd.read_sql_query(sql, db_connection)

# Display the table
display(df6)
```

### Requirement 7: 

Determine the most influential product reviewer for every product sale.  **( Use Python - SQL - Postgres )**

**Criteria for an Influential Reviewer:**
1. The reviewer must have rated the product either 5 (excellent) or 1 (poor).
2. The reviewer must have written a review for the product.
3. The reviewer is considered more influential based on two factors:
   - The number of friends they have.
   - The number of friends with whom they have shared the review.

**Data Ordering**: 
- List the reviewers in descending order based on the count of 'sharedwith' first and then 'friends'.

Example Output Format:

2. Reviewer ID: c1 - SharedWith: 4, Friends: 15

### Suggestions for Requirement 7:

1. **Filtering Reviews**: Start by isolating rows where the review has a rating of either 1 or 5 and where a `reviewid` is specified, indicating a reviewed product.

2. **Non-Empty `friends` and `sharedwith`**: Ensure that both the `friends` and `sharedwith` columns are non-empty, as these are critical for assessing influence.

3. **Unique Customer IDs**: Work under the assumption that each `customerid` is unique for the purpose of this analysis, or ignore `customerid`s that occur more than once in the table.

4. **Parsing Semicolon-separated Lists**: Decide whether to parse the semicolon-separated lists in the `friends` and `sharedwith` columns using SQL functions or Pandas methods.

5. **SQL String Functions**: If using SQL, look into functions like `string_to_array` and `array_length` or `cardinality` for splitting strings into arrays and counting elements.

6. **Pandas String Operations**: In Pandas, utilize `str.split(';')` to split the strings and apply `len` to count the elements for both `friends` and `sharedwith`.

7. **Sorting for Influence**: Sort the data based on the influence metrics, prioritizing reviewers with the highest number of shares and friends.


```python
import pandas as pd

sql = """
    SELECT 
    customerid,
    firstname,
    lastname,
    reviewid,
    friends,
    cardinality(string_to_array(TRIM(friends), ';')) AS num_friends,
    sharedwith,
    cardinality(string_to_array(TRIM(sharedwith), ';')) AS num_shares,
    rating
FROM transactions_log
WHERE rating IN (1, 5)
  AND reviewid IS NOT NULL
  AND friends IS NOT NULL
  AND sharedwith IS NOT NULL
  AND TRIM(reviewid::text) != ''
  AND TRIM(friends::text) != ''
  AND TRIM(sharedwith::text) != ''
ORDER BY num_shares DESC;
"""

# Load directly into a DataFrame using your active psycopg2 connection ('conn')
df7 = pd.read_sql_query(sql, db_connection)

# Display the table
display(df7)
```

### Requirement 8: 
- OnMart has allocated the budget for the next quarter to expand its logistics and supply network by adding a new warehouse in the region that has the highest number of sale orders. Inspect the map below for the list of regions of United States and find the region that has the highest number of orders based on the delivery zip code and the zip code of the distribution center **( Use Python - SQL - Postgres )**
- Your choice to either report based on the four regions shown - West, Midwest, Northeast and South **OR** you can report more granular using the sub-divisions (example: Pacific, Mountain, West North Central, et).


</div>


<br><br>

Source: https://upload.wikimedia.org/wikipedia/commons/f/f1/Census_Regions_and_Division_of_the_United_States.svg

<br><br>
    


    


</div>

```python
import pandas as pd

sql = """
WITH OrderRegions AS (
    SELECT 
        t.orderid,
        t.deliveryzipcode,
        l.distribution_center_id,
        -- Extract the 3-digit zip code prefix and map to the 9 Census Divisions
        CASE 
            -- New England: CT, ME, MA, NH, RI, VT
            WHEN LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '010' AND '069' THEN 'New England'
            
            -- Middle Atlantic: NJ, NY, PA
            WHEN LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '070' AND '196' THEN 'Middle Atlantic'
            
            -- South Atlantic: DE, DC, FL, GA, MD, NC, SC, VA, WV
            WHEN LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '197' AND '349' THEN 'South Atlantic'
            
            -- East South Central: AL, KY, MS, TN
            WHEN LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '350' AND '427' THEN 'East South Central'
            
            -- East North Central: IL, IN, MI, OH, WI
            WHEN LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '430' AND '499' 
              OR LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '530' AND '549'
              OR LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '600' AND '629' THEN 'East North Central'
            
            -- West North Central: IA, KS, MN, MO, NE, ND, SD
            WHEN LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '500' AND '528'
              OR LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '550' AND '588'
              OR LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '630' AND '693' THEN 'West North Central'
            
            -- West South Central: AR, LA, OK, TX
            WHEN LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '700' AND '799' THEN 'West South Central'
            
            -- Mountain: AZ, CO, ID, MT, NV, NM, UT, WY
            WHEN LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '800' AND '898' THEN 'Mountain'
            
            -- Pacific: AK, CA, HI, OR, WA
            WHEN LEFT(LPAD(t.deliveryzipcode::text, 5, '0'), 3) BETWEEN '900' AND '999' THEN 'Pacific'
            
            ELSE 'Unknown Region'
        END AS region
    FROM transactions_log t
    INNER JOIN logistics_supply_chain_network l 
        ON LPAD(t.deliveryzipcode::text, 5, '0') = LPAD(l.zipcode::text, 5, '0')
)
SELECT 
    region,
    COUNT(DISTINCT orderid) AS total_orders
FROM OrderRegions
WHERE region != 'Unknown Region'
GROUP BY region
ORDER BY total_orders DESC
LIMIT 1;
"""

# Load directly into a DataFrame using your active psycopg2 connection ('conn')
df8 = pd.read_sql_query(sql, db_connection)

# Display the table
display(df8)

```

### Requirement 9: 
- To better plan for hiring hourly temporary workers on the peak day of the week (Monday, Tuesday, Wednesday, Thursday, Friday, Satruday or Sunday), OnMart has requested to identify the day with the highest average number of order deliveries in every zip-code **( Use Python - SQL - Postgres )**





</div>

```python
import pandas as pd

sql = """
    WITH DailyDeliveries AS (
    SELECT 
        deliveryzipcode,
        TO_CHAR(actualdeliverydate::date, 'Day') AS day_of_week,
        COUNT(DISTINCT orderid) AS total_deliveries
    FROM transactions_log
    WHERE deliveryzipcode IS NOT NULL 
      AND actualdeliverydate IS NOT NULL
    GROUP BY 
        deliveryzipcode, 
        TO_CHAR(actualdeliverydate::date, 'Day')
),
AverageDeliveries AS (
    SELECT 
        deliveryzipcode,
        TRIM(day_of_week) AS peak_delivery_day,
        AVG(total_deliveries) AS avg_deliveries,
        DENSE_RANK() OVER (
            PARTITION BY deliveryzipcode 
            ORDER BY AVG(total_deliveries) DESC
        ) AS rank
    FROM DailyDeliveries
    GROUP BY deliveryzipcode, TRIM(day_of_week)
)
SELECT 
    deliveryzipcode,
    peak_delivery_day,
    ROUND(avg_deliveries::numeric, 2) AS max_avg_deliveries
FROM AverageDeliveries
WHERE rank = 1
ORDER BY deliveryzipcode;
"""

# Load directly into a DataFrame using your active psycopg2 connection ('conn')
df9 = pd.read_sql_query(sql, db_connection)

# Display the table
display(df9)
```
